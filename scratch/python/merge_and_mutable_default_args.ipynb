```python
def merge(a: list[float], b: list[float]) -> list[float]:
    """Return a new sorted list containing all elements of a and b.

    Assumes a and b are each already in non-decreasing order.
    Does not modify a or b.
    """
    if not a:
        return b
    if not b:
        return a
    id1, id2, hi1, hi2 = 0, 0, len(a), len(b)
    result = []
    while (id1 < hi1) and (id2 < hi2):
        if a[id1] <= b[id2]:
            result.append(a[id1])
            id1 += 1
        else:
            result.append(b[id2])
            id2 += 1
    if id1 < hi1:
        result += a[id1:]  # += in place extend. result.extend(a[id1:]) is in place too.
    elif id2 < hi2:
        result += b[id2:]
    return result
```

```python
assert merge([1, 4, 9], [2, 3, 10]) == [1, 2, 3, 4, 9, 10]
assert merge([], [5, 6]) == [5, 6]
assert merge([], []) == []
```

```python
def add(item, items=[]):
    print("List identity:", id(items))
    items.append(item)
    return items
```
