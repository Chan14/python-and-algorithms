Python Challenge: "Mini Temperature Analyzer"
Scenario:
You have daily temperatures (in °C) for a week. You want to find:

The hottest day
The coldest day
The average temperature
The number of days above 30°C

Data:
temps = [28, 32, 31, 29, 35, 27, 30]

Tasks:
Use Python (and optionally NumPy) to print:

"Hottest day temperature: X°C"
"Coldest day temperature: Y°C"
"Average temperature: Z°C"
"Days above 30°C: N"
Bonus (if you want):
Print "Above 30°C days:" followed by a list of those temperatures
Guidelines:

Keep it simple, work in your phone Python editor / REPL
You can use loops, list comprehension, or NumPy functions
Try to do it all in one short script


```python
temps = [28, 32, 31, 29, 35, 27, 30]
max_temp = max(temps)
min_temp = min(temps)
avg_temp = sum(temps) / len(temps)
above_30 = [i for i in range(len(temps)) if temps[i] > 30]
days_above_30 = len(above_30)
exact_temps = [temps[day] for day in above_30]
exact_days = [day + 1 for day in above_30]
```

```python
import numpy as np

temps = [28, 32, 31, 29, 35, 27, 30]
max_temp = np.max(temps)
min_temp = np.min(temps)
avg_temp = np.mean(temps)

temps = np.asarray(temps)
above_30 = temps > 30
print(above_30)
print(temps[above_30])
print(np.where(above_30)[0] + 1)
```
