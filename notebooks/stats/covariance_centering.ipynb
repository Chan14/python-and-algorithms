```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[10, 4], [-12, 12], [16, 24], [15, 10]])
mu = np.mean(X, axis=0)
D_c = X - mu
print(f"X = \n{X}")
print(f"\nmu = {mu}")
print(f"\nD_c = \n{D_c}")
```

```python
left = np.ones((4, 1))
right = mu.reshape(1, -1)
print(left)
print(right)
print(np.allclose(X - (left @ right), D_c))
```
