**Calibration Problem: Climbing Stairs**

You are climbing a staircase with $n$ steps.
Each move, you can climb either **1 step** or **2 steps**.
How many distinct ways can you reach the top?

_Examples:_

- $n = 2$: 2 ways → (1+1), (2)
- $n = 3$: 3 ways → (1+1+1), (1+2), (2+1)
- $n = 4$: 5 ways → figure this out yourself

_Your task_

Write a Python function:

```python
def climb_stairs(n: int) -> int:
    # your code here
```

_Constraints for this attempt:_

- Use _pure recursion only_ — no loops, no memoization yet. I want to see your instinct.
- Don't look anything up.
- Java-to-Python translation tip: Python has no type declarations in the body, no braces, indentation is the block structure, and `return` works identically.

_One hint about the invariant to reason from before you write a single line:_

Stand at step $n$. You got there from either step $n-1$ or step $n-2$. That's the only two possibilities.
Think about what that means for the number of ways to reach step $n$. Then write the recurrence in a comment first, code second.
Take your time — there's no rush. Show me your reasoning and your code.


````python
# Wrong first attempt
def climb_stairs(n: int) -> int:
    if n < 1:
        return 1
    if n == 1:
        return 1
    return climb_stairs(n - 2) + climb_stairs(n - 1)
````

````python
climb_stairs(1)
````

Claude's feedback

It works — every input from 0 up returns the correct answer. But I'm going to hold you to the invariant-first standard, because "works" and "correct by construction" are two different things, and the gap between them is exactly what DP will punish later.

Here's the tell. Watch what happens if I delete the `n == 1` line from _your_ version:

```python
def climb_stairs(n):
    if n < 1:
        return 1
    return climb_stairs(n - 2) + climb_stairs(n - 1)
```

Trace $n = 1$:

```python
climb_stairs(1)
→ climb_stairs(-1) + climb_stairs(0)
→ 1 + 1
→ 2   ✗   (should be 1)
```

It breaks. So your `n == 1` line is **load-bearing** — the code depends on it. That means it's still the redundant scaffolding I asked you to remove two turns ago; it just changed disguise. It's propping up a base case that is quietly wrong.

**Where's the wrongness?** This line:

```python
    if n < 1:
        return 1
```

This says: _"reaching any step at or below 0 — including step −1, step −2 — counts as 1 way."_ But landing on a negative step is an illegal overshoot. It should be **0 ways**, not 1. Your code never actually reaches a negative _only because_ the `n == 1` guard stops the recursion one step early. Two wrongs propping each other up.

---

Now watch the clean version — negative and zero as **separate** base cases:

```python
def climb_stairs(n):
    if n < 0:      # overshoot: illegal, 0 ways
        return 0
    if n == 0:     # empty path: 1 way
        return 1
    return climb_stairs(n - 2) + climb_stairs(n - 1)
```

Trace $n = 1$ with _no_ special case for it:

```python
climb_stairs(1)
→ climb_stairs(-1) + climb_stairs(0)
→ 0 + 1
→ 1   ✓
```

The recurrence now produces the right answer for $n = 1$ **on its own**. That's the proof the base cases are sound: once they're correct, every special case dissolves. `n == 1` is deletable because it was never really a base case — it was a patch.

---

**The two invariants, stated plainly:**

- $n < 0$ → 0. There is no way to legally land past the top.
- $n = 0$ → 1. There is exactly one empty path: take no steps.

Everything above 0 is pure recurrence.

---

This distinction — _correct by accident_ versus _correct because the base cases are individually true_ — is the single most valuable habit for the RL material coming in Course 3. The Bellman equation is a recurrence with base cases, and if your terminal-state value is wrong, the whole value function is silently poisoned upward, exactly like your negative-step case would have been.

You've got the recurrence nailed. Adopt the clean two-base-case version, confirm you see _why_ `n == 1` disappears, and then we move to the thing that makes this actually **dynamic** programming — because right now this function has an exponential-time problem hiding in it. Ready when you are.


````python
# The right solution
def climb_stairs(n: int) -> int:
    if n < 0:
        return 0
    if n == 0:
        return 1
    return climb_stairs(n - 2) + climb_stairs(n - 1)
````
