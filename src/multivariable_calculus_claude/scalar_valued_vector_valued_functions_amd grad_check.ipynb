```python
import numpy as np
```

```python
# Scalar value functions and vector value functions.
def f(w):  # scalar-valued: R^n -> R
    return np.sum(w**2)  # many numbers in, ONE number out


def g(w):  # vector-valued: R^n -> R^n
    return w**2  # many numbers in, many numbers out


w = np.array([1.0, 2.0, 3.0])
print(f(w))  # 14.0  (a single number)
print(g(w))  # [1. 4. 9.]  (a vector)
```

```python
# Gradient and gradient checking
def f(w):
    return np.sum(w**2)


def analytic_grad(w):
    return 2 * w  # our hand-derived result


def numeric_grad(f, w, eps=1e-5):
    g = np.zeros_like(w)
    for i in range(len(w)):  # one knob at a time (the partial-derivative idea)
        up, down = w.copy(), w.copy()
        up[i] += eps
        down[i] -= eps
        g[i] = (f(up) - f(down)) / (2 * eps)  # central difference
    return g


w = np.array([3.0, 4.0])
print(analytic_grad(w))  # [6. 8.]
print(numeric_grad(f, w))  # ~[6. 8.]
```
